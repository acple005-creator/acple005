```python
# Q6 100초를 "%d 분 %d 초 입니다." 로 표기 하시오. //을 사용하여 소수점을 제거

seconde = 100
sec_min= seconde // 60
sec_sec= seconde % 60
print("100초는 %d 분 %d 초 입니다." % (sec_min, sec_sec))
```

```python
# Q7 800원 에서 500원을 제거하여 "100원의 갯수를 구하시오"
# 마찬가지로 total//100을 사용하여 소수점 제거하여 십원짜리 또한 제거

total_money = 800
removed_500 = 500
total = total_money - removed_500
print("100원의 개수는 %d 개입니다." % (total//100))
```

```python
# Q8 이름 데이터 처리
data = "이유덕,이재영,권종표,이재영,박민호,강상희,이재영,김지원,최승혁,이성연,박영서,박민호,전경헌,송정환,김재성,이유덕,전경헌"
data_split = data.split(',')
# 김 , 이씨 명수를 구하시오. - count'김' '이'
print(data.count('김'),data.count('이'))
# 이재영 이름을 사용하는 명수를 구하시오.
print(data.count('이재영'))
# 중복을 제거하시오 - 함수로 변경 함수는 중복 존재 x set()
print(set(data_split))
# 중복을 제거한 인원을 이름순으로 나열하세요. - 이름순 나열함수 sorted(a)
print(sorted(set(data_split)))
```

```python
# 김씨와 이씨 명수 구하기 성만 따로 리스트 만들어서 count 하기
# 리스트 내포 사용 표현 [표현식 for 항목 in 반복_가능_객체 if 조건문]
data = "이유덕,이재영,권종표,이재영,박민호,강상희,이재영,김지원,최승혁,이성연,박영서,박민호,전경헌,송정환,김재성,이유덕,전경헌"
data_split = data.split(',')
l_data = [i[0] for i in data_split]
print(l_data.count('김'),l_data.count('이'))
```

```python
# Q9 점수 등급 분류
score = [80, 60, 70, 50, 100, 95, 40, 75, 90, 65, 90, 100]


grade = [
    '수 %d'%s if s >= 90 else
    '우 %d'%s if s >= 80 else
    '미 %d'%s if s >= 70 else
    '재시험 %d'%s
    for s in score
]

print(grade)
```
