```python
while True:
    n = int(input("계산할 숫자를 입력해주세요. : "))
    while True:
        if n == 1: 
            print("숫자가 1이 됨으로 계산이 종료됩니다.")
            break

        if n % 2 == 0: 
            n = n / 2
            print("짝수입니다. n을 2로 나눈 값 =", n)

        else:  
            n = n * 3 + 1
            print("홀수입니다. n * 4 =", n)
    break
```
