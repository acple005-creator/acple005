```python
start = input("369 게임을 시작할까요? (네/yes 입력 시 시작): ")

if start.lower() in ['네', 'yes']:
    n = 100
    for i in range(1, n + 1):
        s = str(i)
        count = 0
        for x in s:
            if x in ('3', '6', '9'):
                count += 1

        if count == 0:
            print(i, end=' ')
        else:
            print(count * '짝', end=' ')
else:
    print("게임을 시작하지 않습니다.")
```

```python
while True:
    n = int(input("계산할 숫자를 입력해주세요. : "))
    while True:
        if n == 1: 
            print("숫자가 1이 됨으로 계산이 종료됩니다.")
            break

        if n % 2 == 0: 
            n = n / 2
            print("짝수입니다. n을 2로 나눈 값 =", n)

        else:  
            n = n * 3 + 1
            print("홀수입니다. n * 4 =", n)
    break
```
