```python
coffee = 10
while True:
    money = int(input("돈을 넣어 주세요: "))
    if money == 300:
        print("커피를 줍니다.")
        coffee = coffee -1
    elif money > 300:
        print("거스름돈 %d를 주고 커피를 줍니다." % (money -300))
        coffee = coffee -1
    else:
        print("돈을 다시 돌려주고 커피를 주지 않습니다.")
        print("남은 커피의 양은 %d개 입니다." % coffee)
    if coffee == 0:
        print("커피가 다 떨어졌습니다. 판매를 중지 합니다.")
        break
```

```python
coffee = 10  # Default coffee count
while True:
    try:
        money = int(input("돈을 넣어 주세요: "))
        quantity = int(input("구매할 커피 개수를 입력하세요: "))

        if quantity > coffee:
            print("남은 커피가 부족합니다. 현재 남은 커피: %d개" % coffee)
            print("돈을 다시 돌려주고 커피를 주지 않습니다.")
        else:
            total_price = quantity * 300
            if money >= total_price:
                change = money - total_price
                print("커피 %d잔을 줍니다." % quantity)
                if change > 0:
                    print("거스름돈 %d억을 드립니다." % change)
                coffee -= quantity
            else:
                print("돈이 부족합니다. 필요한 금액: %d원" % total_price)
                print("돈을 다시 돌려주고 커피를 주지 않습니다.")

        print("남은 커피의 양은 %d개 입니다." % coffee)

        if coffee == 0:
            print("커피가 다 떨어졌습니다. 판매를 중지 합니다.")
            break
    except ValueError:
        print("잘못된 입력입니다. 숫자를 입력해주세요.")
```
